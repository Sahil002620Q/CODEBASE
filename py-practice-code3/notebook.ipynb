# SET in py

 can be modefied , will remove repeative

```python
#set will remove repeative element
set = {11,22,11,4355,'e'}
print(set)
for s in set:
    print(s)
# print(set[1]) will not work
set.discard(9090203849383) #no error if value is not in set due to discard
print(11 in set) #check is that val is in arr,set,list etc
```

```python
#set union     //merge teo set igonore repeative elements
set_1 = {1,2,3}
set_2 = {3,4,5}
set_unioned = set_1 | set_2
print(set_unioned)
```

```python
#set intersection    //finds common
set_1 = {1,2,3}
set_2 = {3,4,5}
set_intersected = set_1 & set_2
print(set_intersected)
```

```python
#set diff     // subtract all elements of 2 from 1 and print all remaining elemts in 1 (rest elements of 2 will be ignored which are not in 1 ) 
set_1 = {1,2,3}
set_2 = {3,4,5}
set_diff = set_1 - set_2
print(set_diff)
```

```python
#set symmetry    //finds non common or remove common
set_1 = {1,2,3}
set_2 = {3,4,5}
set_uncommon = set_1 ^ set_2
print(set_uncommon)
```

# intracting with database using sqlite3

```python
import sqlite3

inject = [
    (1,"sahil"),
    (2,"sam")
]

#if not exist , use it if using it multiple time so that it doesn't create multiple time

with sqlite3.connect("first.db") as connection:
    cursor = connection.cursor()
    cursor.execute("""
                   CREATE TABLE IF NOT EXISTS first (
                   sno INTEGER,
                   name TEXT)
                   """)

    for sno,name in inject:
        cursor.execute("INSERT INTO first VALUES (?,?)",(sno,name))
```
