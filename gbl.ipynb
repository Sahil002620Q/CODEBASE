Text Type     :	str
Numeric Types :	int, float, complex
Sequence Types:	list, tuple, range
Mapping Type  :	dict
Set Types     :	set, frozenset
Boolean Type  :	bool
Binary Types  :	bytes, bytearray, memoryview

```python
#Data type in python
print(type(10))
print(type(10e10))
print(type('sahil'))
print(type((1+1j))) #complex
print(type([1,2.0,3,"sahil"]))
print(type((10,20.0,"sahil")))
print(type({"key":"value"}))
print(type({1,2,3,3}))
print(type(True))

#unique

print(type(b"hello"))
print(type(range(20)))
print(type(bytearray(1)))
print(type(frozenset({1,3,2,3})))
print(type(memoryview(bytes(3))))
print(type(None))
```
